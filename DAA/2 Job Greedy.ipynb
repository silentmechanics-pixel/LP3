```python
n = int(input("Enter no of jobs: "))

jobs = []

for i in range (n):
    jobname = input("Enter name: ")
    jobdeadline = int(input("Enter deadline: "))
    jobprofit = int(input("Enter profit: "))

    jobs.append([jobname, jobdeadline, jobprofit])

#sort wrt profit
for i in range (n):
    for j in range (i+1, n):
        if jobs[i][2] < jobs[j][2]:
            jobs[i], jobs[j] = jobs[j], jobs[i]

#find max deadline
maxdeadline = 0;

for i in range (n):
    if jobs[i][1] > maxdeadline:
        maxdeadline = jobs[i][1]

#create empty slots
slots = [-1] * maxdeadline

#profit variable 
totalprofit = 0

#scheduling
for i in range(n):
    deadline = jobs[i][1]

    for j in range(deadline-1, -1, -1):
        if slots[j] == -1:
            slots[j] = i
            totalprofit += jobs[i][2]
            break

for i in range(maxdeadline):
    if slots[i] != -1:
        print(jobs[slots[i]][0])

print("Total profit: ", totalprofit)
```

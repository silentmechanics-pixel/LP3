```python
n = int(input("Enter no of items: "))

items = []

for i in range(n):
    weight = int(input("Enter weight: "))
    profit = int(input("Enter profit: "))
    ratio = profit / weight

    items.append([weight, profit, ratio])

capacity = int(input("Enter knapsack capacity"))

#sorting acc to ratio
for i in range(n):
    for j in range(i+1, n):
        if items[i][2] < items[j][2]:
            items[i], items[j] = items[j], items[i]

totalprofit = 0

#knapsacking
for i in range (n):
    if items[i][0] <= capacity:
        totalprofit += items[i][1]
        capacity = capacity-items[i][0]

    else:
        fraction = capacity / items[i][0]
        totalprofit += fraction * items[i][1]
        break

print("Maximum profit: ", totalprofit)
```
